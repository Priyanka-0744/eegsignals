```python
import scipy.io
import matplotlib.pyplot as plt

data=scipy.io.loadmat(r"C:\users\krish\Downloads\SEED_Dataset\Preprocessed_EEG/1_20131027.mat")
eeg = data['djc_eeg1']


sampling_data = 250 # Hz
time = [i / sampling_data for i in range(eeg.shape[1])] # in seconds


plt.figure(figsize=(12, 6))

for k in range(3):
    plt.plot(time, eeg[k], label=f'Channel {k+1}')

plt.title('ECG Signal - Trial 1 (Channels 3)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude (mV)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
